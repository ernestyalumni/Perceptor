```python
import sympy
import sympy.physics.units as units
from sympy import Eq, solve
```

```python
V_batteries_total = 18.8 * units.V
R_1 = 10 * 10**3 * units.ohm
# Desired Voltage:
V_desired = 12 * units.V
# Desired current
I_desired = 0.16 * units.amperes
```

```python
# See https://omz-software.com/pythonista/sympy/modules/physics/units.html
units.find_unit('ampere')
```

```python
R_total_desired = V_batteries_total / I_desired
print("Desired total resistance: ", R_total_desired)
print("Desired proportion: ", V_desired / I_desired)
print("Other resistance value: ", R_total_desired - V_desired / I_desired)
```

Recall resistors adding in parallel:

```python
R1_value, R2_value, Rparallel_value = sympy.symbols(['R1_value', 'R2_value', 'Rparallel_value'])
R1 = R1_value * units.ohm
R2 = R2_value * units.ohm
Rparallel = Rparallel_value * units.ohm
resistors_add_in_parallel = Eq(1.0 / Rparallel, 1.0 / R1 + 1.0 /R2)
```

```python
Rparallel_solved = solve(resistors_add_in_parallel, Rparallel)
print(type(Rparallel_solved))
Rparallel_solved = Rparallel_solved[0]
print(Rparallel_solved)
```

```python
Rparallel_solved.subs(R1_value, 10000).subs(R2_value, 10000)
```
